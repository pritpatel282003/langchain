```python
import os
import psycopg2
from psycopg2.extras import RealDictCursor
from langchain_openai import AzureOpenAIEmbeddings
from dotenv import load_dotenv
import logging

# Load .env config
load_dotenv()

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

model_name = os.getenv('AZURE_OPENAI_MODEL')
api_endpoint = os.getenv('AZURE_ENDPOINT')
api_version = os.getenv('AZURE_API_VERSION')
api_key = os.getenv('AZURE_API_KEY')

# Initialize embeddings
embeddings = AzureOpenAIEmbeddings(
    model=model_name,
    openai_api_type="azure",
    azure_endpoint=api_endpoint,
    openai_api_version=api_version,
    openai_api_key=api_key,
)

# PostgreSQL connection string
connection_string = "postgresql://talent31stage:centralIndia%402025@talent31-stage-db-centralindia.postgres.database.azure.com:5432/talent31-db-stage-ind?sslmode=require"


def hybrid_search(query, designation=None, skills=None, min_experience=None, k=10):
    """
    Performs hybrid search combining semantic similarity with keyword matching boosts.
    """
    try:
        conn = psycopg2.connect(connection_string, cursor_factory=RealDictCursor)
        cursor = conn.cursor()

        # Convert query text to embedding
        query_embedding = embeddings.embed_query(query)
        
        # Debug: Check embedding dimension
        logger.info(f"Query embedding dimension: {len(query_embedding)}")
        
        # Convert embedding to PostgreSQL array format
        query_vector_str = "[" + ",".join([str(x) for x in query_embedding]) + "]"

        # Start building the SQL query and parameters
        base_sql = """
        SELECT 
            c."Id",
            c."UserId",
            c."DesignationId",
            c."ExperienceYears",
            c."City",
            c."State",
            c."PostalCode",
            c."Summary",
            c."ExpectedCTCFrom",
            c."ExpectedCTCTo",
            c."CurrentCTC",
            c."AdditionalSkills",
            1 - (c."Embedding" <=> %s::vector) AS similarity_score,
        """
        
        # Initialize parameters list with the query embedding (used twice)
        params = [query_vector_str, query_vector_str]
        
        # Build boost conditions dynamically
        boost_conditions = ["(1 - (c.\"Embedding\" <=> %s::vector)) * 0.6"]  # Base semantic similarity (60%)
        
        # Add designation boost if provided
        if designation:
            boost_conditions.append("CASE WHEN LOWER(c.\"Summary\") LIKE LOWER(%s) THEN 0.2 ELSE 0 END")
            params.append(f"%{designation.lower()}%")
        
        # Add skills boost if provided
        if skills:
            boost_conditions.append("CASE WHEN LOWER(c.\"AdditionalSkills\") LIKE LOWER(%s) THEN 0.15 ELSE 0 END")
            params.append(f"%{skills.lower()}%")
        
        # Add experience boost if provided
        if min_experience and min_experience > 0:
            boost_conditions.append("CASE WHEN c.\"ExperienceYears\" >= %s THEN 0.15 ELSE 0 END")
            params.append(min_experience)
        
        # Combine all boost conditions
        boost_sql = " + ".join(boost_conditions)
        
        # Complete the SQL query
        complete_sql = base_sql + f"""
            ({boost_sql}) AS boosted_score
        FROM "Candidate" c
        WHERE c."Embedding" IS NOT NULL
        ORDER BY boosted_score DESC
        LIMIT %s;
        """
        
        # Add the limit parameter
        params.append(k)
        
        # Debug: Print SQL and parameter count
        logger.info(f"SQL placeholders count: {complete_sql.count('%s')}")
        logger.info(f"Parameters count: {len(params)}")
        
        # Execute the query
        cursor.execute(complete_sql, params)
        results = cursor.fetchall()
        
        logger.info(f"Found {len(results)} candidates")
        
        return results
        
    except Exception as e:
        logger.error(f"Error in hybrid_search: {str(e)}")
        raise
    finally:
        if 'conn' in locals():
            conn.close()


def test_basic_search(query, k=5):
    """
    Simple semantic search without boosts for testing.
    """
    try:
        conn = psycopg2.connect(connection_string, cursor_factory=RealDictCursor)
        cursor = conn.cursor()
        
        # Get query embedding
        query_embedding = embeddings.embed_query(query)
        query_vector_str = "[" + ",".join([str(x) for x in query_embedding]) + "]"
        
        sql = """
        SELECT 
            c."Id",
            c."Summary",
            c."AdditionalSkills",
            c."ExperienceYears",
            1 - (c."Embedding" <=> %s::vector) AS similarity_score
        FROM "Candidate" c
        WHERE c."Embedding" IS NOT NULL
        ORDER BY similarity_score DESC
        LIMIT %s;
        """
        
        cursor.execute(sql, [query_vector_str, k])
        results = cursor.fetchall()
        
        return results
        
    except Exception as e:
        logger.error(f"Error in test_basic_search: {str(e)}")
        raise
    finally:
        if 'conn' in locals():
            conn.close()


if __name__ == "__main__":
    print("=== Testing Basic Search First ===")
    try:
        query = "full stack developer python aws"
        basic_results = test_basic_search(query, k=3)
        

            
    except Exception as e:
        print(f"Basic search failed: {e}")
    
    print("\n" + "="*50)
    print("=== Testing Hybrid Search ===")
    try:
        query = """We are looking for an Android Developer (India-based) with 3–5 years of experience.
Must-have skills: DART, Android Studio, Java & Kotlin,RESTful API integration, android jet pack ,SQLite Room Database, Firebase (Auth, Messaging, Analytics), Git / GitHub / GitLab ,MVVM, MVP, Figma , Material UI, Postman"""
        designation = "Android Developer"
        skills = "DART, Android Studio, Java & Kotlin,RESTful API integration, android jet pack ,SQLite Room Database, Firebase (Auth, Messaging, Analytics), Git / GitHub / GitLab ,MVVM, MVP, Figma , Material UI, Postman"
        min_experience = 3

        candidates = hybrid_search(query, designation, skills, min_experience, k=10)

        print(f"Hybrid search found {len(candidates)} results")
        for i, c in enumerate(candidates, 1):
            print(f"\n--- Hybrid Candidate {i} ---")
            print(f"ID: {c['Id']}")
            print(f"Name: {c.get('Name', 'N/A')}")
            print(f"Projects: {c.get('Projects', 'N/A')}")
            print(f"Boosted Score: {c['boosted_score']:.3f}")
            print(f"Similarity Score: {c['similarity_score']:.3f}")
            print(f"Experience: {c['ExperienceYears']} years")
            print(f"Skills: {c['AdditionalSkills']}")
            print(f"Summary: {c['Summary'][:150] if c['Summary'] else 'No summary'}...")
            
    except Exception as e:
        print(f"Hybrid search failed: {e}")
        import traceback
        traceback.print_exc()
```

```python
import os
import psycopg2
from psycopg2.extras import RealDictCursor
from langchain_openai import AzureOpenAIEmbeddings
from langchain_community.utilities import WikipediaAPIWrapper
from dotenv import load_dotenv
import logging
import re

# Load .env config
load_dotenv()

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

model_name = os.getenv('AZURE_OPENAI_MODEL')
api_endpoint = os.getenv('AZURE_ENDPOINT')
api_version = os.getenv('AZURE_API_VERSION')
api_key = os.getenv('AZURE_API_KEY')

# Initialize embeddings
embeddings = AzureOpenAIEmbeddings(
    model=model_name,
    openai_api_type="azure",
    azure_endpoint=api_endpoint,
    openai_api_version=api_version,
    openai_api_key=api_key,
)

# Initialize Wikipedia wrapper
wikipedia = WikipediaAPIWrapper()

# PostgreSQL connection string
connection_string = "postgresql://talent31stage:centralIndia%402025@talent31-stage-db-centralindia.postgres.database.azure.com:5432/talent31-db-stage-ind?sslmode=require"


def normalize_skills(skills_text):
    """
    Normalize skills text for better matching by handling common variations.
    """
    if not skills_text:
        return ""
    
    # Convert to lowercase
    normalized = skills_text.lower()
    
    # Define skill normalization mappings
    skill_mappings = {
        'node.js': ['nodejs', 'node js', 'node-js'],
        'react.js': ['reactjs', 'react js', 'react-js'],
        'vue.js': ['vuejs', 'vue js', 'vue-js'],
        'angular.js': ['angularjs', 'angular js', 'angular-js'],
        'next.js': ['nextjs', 'next js', 'next-js'],
        'express.js': ['expressjs', 'express js', 'express-js'],
        'c++': ['cpp', 'c plus plus'],
        'c#': ['csharp', 'c sharp'],
        'postgresql': ['postgres', 'psql'],
        'mongodb': ['mongo db', 'mongo'],
        'javascript': ['js'],
        'typescript': ['ts'],
        'artificial intelligence': ['ai'],
        'machine learning': ['ml'],
        'natural language processing': ['nlp'],
        'computer vision': ['cv'],
        'amazon web services': ['aws'],
        'google cloud platform': ['gcp'],
        'microsoft azure': ['azure'],
        'continuous integration': ['ci'],
        'continuous deployment': ['cd'],
        'devops': ['dev ops', 'dev-ops'],
        'restful': ['rest'],
        'graphql': ['graph ql', 'graph-ql']
    }
    
    # Apply normalization mappings
    for canonical, variations in skill_mappings.items():
        for variation in variations:
            normalized = normalized.replace(variation, canonical)
    
    return normalized


def get_wikipedia_context(query, max_docs=2):
    """
    Get relevant Wikipedia context for the query to enhance understanding.
    """
    try:
        # Extract key terms from query for Wikipedia search
        key_terms = []
        tech_terms = ['python', 'java', 'nodejs', 'aws', 'docker', 'kubernetes', 'postgresql', 'mysql', 'redis', 'terraform', 'elasticsearch']
        
        query_lower = query.lower()
        for term in tech_terms:
            if term in query_lower:
                key_terms.append(term)
        
        if not key_terms:
            return ""
        
        context = ""
        for term in key_terms[:max_docs]:  # Limit to avoid too much context
            try:
                wiki_result = wikipedia.run(term)
                # Take first 200 characters to avoid overwhelming the embedding
                context += f"{term}: {wiki_result[:200]}... "
            except Exception as e:
                logger.warning(f"Failed to get Wikipedia context for {term}: {e}")
                continue
        
        return context
        
    except Exception as e:
        logger.warning(f"Failed to get Wikipedia context: {e}")
        return ""


def enhanced_hybrid_search(query, designation=None, skills=None, min_experience=None, education_keywords=None, preferred_state=None, preferred_country=None, k=10, use_wikipedia=True):
    """
    Performs enhanced hybrid search with education boost, Wikipedia context, location preferences, and better skill matching.
    
    Args:
        query: Main search query
        designation: Target job designation
        skills: Required skills (comma-separated string)
        min_experience: Minimum years of experience
        education_keywords: Education-related keywords (e.g., "Computer Science", "Engineering", "Masters")
        preferred_state: Preferred state/region for candidates
        preferred_country: Preferred country for candidates
        k: Number of results to return
        use_wikipedia: Whether to enhance query with Wikipedia context
    """
    try:
        conn = psycopg2.connect(connection_string, cursor_factory=RealDictCursor)
        cursor = conn.cursor()

        # Enhance query with Wikipedia context if enabled
        enhanced_query = query
        if use_wikipedia:
            wiki_context = get_wikipedia_context(query)
            if wiki_context:
                enhanced_query = f"{query} {wiki_context}"
                logger.info("Enhanced query with Wikipedia context")

        # Convert enhanced query text to embedding
        query_embedding = embeddings.embed_query(enhanced_query)
        
        # Debug: Check embedding dimension
        logger.info(f"Query embedding dimension: {len(query_embedding)}")
        
        # Convert embedding to PostgreSQL array format
        query_vector_str = "[" + ",".join([str(x) for x in query_embedding]) + "]"

        # Start building the SQL query with education join and additional tables
        base_sql = """
        SELECT 
            c."Id",
            c."UserId",
            u."FirstName",
            u."LastName",
            CONCAT(u."FirstName", ' ', u."LastName") as "Name",
            c."DesignationId",
            d."Name" as "DesignationName",
            c."ExperienceYears",
            c."ExperienceMonths",
            c."City",
            c."State",
            c."PostalCode",
            country."Name" as "CountryName",
            c."Summary",
            c."ExpectedCTCFrom",
            c."ExpectedCTCTo",
            c."CurrentCTC",
            c."AdditionalSkills",
            c."LinkedInUrl",
            c."GitHubUrl",
            c."WebsiteUrl",
            c."NoticePeriod",
            c."IsOpenToWork",
            STRING_AGG(DISTINCT ce."Degree", ', ') as "Degrees",
            STRING_AGG(DISTINCT ce."Institution", ', ') as "Institutions",
            STRING_AGG(DISTINCT proj."Description", ' | ') as "ProjectDescriptions",
            STRING_AGG(DISTINCT exp."Description", ' | ') as "ExperienceDescriptions",
            1 - (c."Embedding" <=> %s::vector) AS similarity_score,
        """
        
        # Initialize parameters list with the query embedding (used twice)
        params = [query_vector_str, query_vector_str]
        
        # Build boost conditions dynamically
        boost_conditions = ["(1 - (c.\"Embedding\" <=> %s::vector)) * 0.4"]  # Base semantic similarity (40%)
        
        # Add designation boost if provided
        if designation:
            boost_conditions.append("CASE WHEN LOWER(c.\"Summary\") LIKE LOWER(%s) OR LOWER(d.\"Name\") LIKE LOWER(%s) THEN 0.2 ELSE 0 END")
            params.extend([f"%{designation.lower()}%", f"%{designation.lower()}%"])
        
        # Enhanced skills boost with normalization
        if skills:
            normalized_skills = normalize_skills(skills)
            skills_list = [skill.strip() for skill in normalized_skills.split(',')]
            
            # Create multiple skill matching conditions
            skill_conditions = []
            for skill in skills_list[:10]:  # Limit to first 10 skills to avoid too many params
                if skill:
                    skill_conditions.append("LOWER(c.\"AdditionalSkills\") LIKE LOWER(%s)")
                    params.append(f"%{skill}%")
            
            if skill_conditions:
                skills_boost = f"(CASE WHEN {' OR '.join(skill_conditions)} THEN 0.25 ELSE 0 END)"
                boost_conditions.append(skills_boost)
        
        # Add experience boost if provided
        if min_experience and min_experience > 0:
            boost_conditions.append("CASE WHEN c.\"ExperienceYears\" >= %s THEN 0.1 ELSE 0 END")
            params.append(min_experience)
        
        # Add project description boost
        if skills:
            normalized_skills_for_project = normalize_skills(skills)
            project_skill_conditions = []
            for skill in normalized_skills_for_project.split(',')[:5]:  # Limit to 5 skills for projects
                skill = skill.strip()
                if skill:
                    project_skill_conditions.append("LOWER(proj.\"Description\") LIKE LOWER(%s)")
                    params.append(f"%{skill}%")
            
            if project_skill_conditions:
                project_boost = f"(MAX(CASE WHEN {' OR '.join(project_skill_conditions)} THEN 0.08 ELSE 0 END))"
                boost_conditions.append(project_boost)
        
        # Add experience description boost
        if skills:
            exp_skill_conditions = []
            for skill in normalized_skills_for_project.split(',')[:5]:  # Limit to 5 skills for experience
                skill = skill.strip()
                if skill:
                    exp_skill_conditions.append("LOWER(exp.\"Description\") LIKE LOWER(%s)")
                    params.append(f"%{skill}%")
            
            if exp_skill_conditions:
                exp_boost = f"(MAX(CASE WHEN {' OR '.join(exp_skill_conditions)} THEN 0.07 ELSE 0 END))"
                boost_conditions.append(exp_boost)
        
        # Add state boost (location preference)
        if preferred_state:
            boost_conditions.append("CASE WHEN LOWER(c.\"State\") LIKE LOWER(%s) THEN 0.08 ELSE 0 END")
            params.append(f"%{preferred_state.lower()}%")
        
        # Add country boost (location preference)
        if preferred_country:
            boost_conditions.append("CASE WHEN LOWER(country.\"Name\") LIKE LOWER(%s) THEN 0.08 ELSE 0 END")
            params.append(f"%{preferred_country.lower()}%")
        
        # Add education boost if provided (using MAX to handle GROUP BY)
        if education_keywords:
            education_conditions = []
            for keyword in education_keywords.split(','):
                keyword = keyword.strip()
                if keyword:
                    education_conditions.append("(LOWER(ce.\"Degree\") LIKE LOWER(%s) OR LOWER(ce.\"Institution\") LIKE LOWER(%s))")
                    params.extend([f"%{keyword.lower()}%", f"%{keyword.lower()}%"])
            
            if education_conditions:
                education_boost = f"(MAX(CASE WHEN {' OR '.join(education_conditions)} THEN 0.15 ELSE 0 END))"
                boost_conditions.append(education_boost)
        
        # Combine all boost conditions
        boost_sql = " + ".join(boost_conditions)
        
        # Complete the SQL query with joins including all related tables
        complete_sql = base_sql + f"""
            ({boost_sql}) AS boosted_score
        FROM "Candidate" c
        LEFT JOIN "User" u ON c."UserId" = u."Id"
        LEFT JOIN "Designation" d ON c."DesignationId" = d."Id"
        LEFT JOIN "Country" country ON c."CountryId" = country."Id"
        LEFT JOIN "CandidateEducation" ce ON c."Id" = ce."CandidateId"
        LEFT JOIN "CandidateProject" proj ON c."Id" = proj."CandidateId"
        LEFT JOIN "CandidateExperience" exp ON c."Id" = exp."CandidateId"
        WHERE c."Embedding" IS NOT NULL
        GROUP BY c."Id", c."UserId", u."FirstName", u."LastName", c."DesignationId", 
                 d."Name", country."Name", c."ExperienceYears", c."ExperienceMonths", c."City", c."State", 
                 c."PostalCode", c."Summary", c."ExpectedCTCFrom", c."ExpectedCTCTo", 
                 c."CurrentCTC", c."AdditionalSkills", c."LinkedInUrl", c."GitHubUrl", 
                 c."WebsiteUrl", c."NoticePeriod", c."IsOpenToWork", c."Embedding"
        ORDER BY boosted_score DESC
        LIMIT %s;
        """
        
        # Add the limit parameter
        params.append(k)
        
        # Debug: Print SQL and parameter count
        logger.info(f"SQL placeholders count: {complete_sql.count('%s')}")
        logger.info(f"Parameters count: {len(params)}")
        
        # Execute the query
        cursor.execute(complete_sql, params)
        results = cursor.fetchall()
        
        logger.info(f"Found {len(results)} candidates")
        
        return results
        
    except Exception as e:
        logger.error(f"Error in enhanced_hybrid_search: {str(e)}")
        raise
    finally:
        if 'conn' in locals():
            conn.close()


# Backward compatibility - keep the original function name with location support
def hybrid_search(query, designation=None, skills=None, min_experience=None, preferred_state=None, preferred_country=None, k=10):
    """
    Original hybrid search function for backward compatibility with location support.
    """
    return enhanced_hybrid_search(query, designation, skills, min_experience, 
                                education_keywords=None, preferred_state=preferred_state, 
                                preferred_country=preferred_country, k=k, use_wikipedia=False)


if __name__ == "__main__":
    print("=== Testing Enhanced Hybrid Search ===")
    try:
        query = """We are looking for a Backend + AWS Developer (India-based) with 3–5 years of experience.
Must-have skills: Python/Java/Node.js, AWS (EC2, S3, Lambda, RDS, CloudFormation), RESTful APIs, Microservices, Docker, Kubernetes, PostgreSQL/MySQL, Redis, Git/GitHub, CI/CD pipelines, Terraform, API Gateway, CloudWatch, ElasticSearch, Message queues (SQS/SNS), Linux/Unix"""
        
        designation = "Backend Developer"
        skills = "Python, Java, Node.js, AWS, EC2, S3, Lambda, RDS, CloudFormation, RESTful APIs, Microservices, Docker, Kubernetes, PostgreSQL, MySQL, Redis, Git, CI/CD, Terraform, API Gateway, CloudWatch, ElasticSearch, SQS, SNS, Linux"
        min_experience = 3
        education_keywords = "Computer Science, Engineering, Masters, Bachelor"
        preferred_state = "Karnataka"
        preferred_country = "India"

        # Test with Wikipedia enhancement and location preferences
        print("\n=== Enhanced Search with Wikipedia and Location ===")
        candidates = enhanced_hybrid_search(
            query=query, 
            designation=designation, 
            skills=skills, 
            min_experience=min_experience,
            education_keywords=education_keywords,
            preferred_state=preferred_state,
            preferred_country=preferred_country,
            k=10,
            use_wikipedia=True
        )

        print(f"Enhanced hybrid search found {len(candidates)} results")
        for i, c in enumerate(candidates, 1):
            print(f"\n--- Enhanced Candidate {i} ---")
            print(f"ID: {c['Id']}")
            print(f"Name: {c.get('Name', 'N/A')}")
            print(f"Designation: {c.get('DesignationName', 'N/A')}")
            print(f"Boosted Score: {c['boosted_score']:.3f}")
            print(f"Similarity Score: {c['similarity_score']:.3f}")
            print(f"Experience: {c['ExperienceYears']} years {c.get('ExperienceMonths', 0)} months")
            print(f"Location: {c.get('City', 'N/A')}, {c.get('State', 'N/A')}, {c.get('CountryName', 'N/A')}")
            print(f"Education: {c.get('Degrees', 'N/A')} from {c.get('Institutions', 'N/A')}")
            print(f"Expected CTC: {c.get('ExpectedCTCFrom', 'N/A')} - {c.get('ExpectedCTCTo', 'N/A')}")
            print(f"Current CTC: {c.get('CurrentCTC', 'N/A')}")
            print(f"Notice Period: {c.get('NoticePeriod', 'N/A')}")
            print(f"Open to Work: {c.get('IsOpenToWork', 'N/A')}")
            print(f"Skills: {c.get('AdditionalSkills', 'No skills listed')}")
            print(f"Summary: {c['Summary'][:200] if c.get('Summary') else 'No summary'}...")
            
            # Display project descriptions if available
            if c.get('ProjectDescriptions'):
                print(f"Project Experience: {c['ProjectDescriptions'][:200]}...")
            
            # Display work experience descriptions if available  
            if c.get('ExperienceDescriptions'):
                print(f"Work Experience: {c['ExperienceDescriptions'][:200]}...")
                
            if c.get('LinkedInUrl'):
                print(f"LinkedIn: {c['LinkedInUrl']}")
            if c.get('GitHubUrl'):
                print(f"GitHub: {c['GitHubUrl']}")
            if c.get('WebsiteUrl'):
                print(f"Website: {c['WebsiteUrl']}")

        # Test skill normalization
        print("\n=== Testing Skill Normalization ===")
        test_skills = "Node.js, React.js, C++, PostgreSQL, MongoDB, AWS"
        normalized = normalize_skills(test_skills)
        print(f"Original: {test_skills}")
        print(f"Normalized: {normalized}")
            
    except Exception as e:
        print(f"Enhanced hybrid search failed: {e}")
        import traceback
        traceback.print_exc()
```

```python
import os
import psycopg2
from psycopg2.extras import RealDictCursor
from langchain_openai import AzureOpenAIEmbeddings
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
from dotenv import load_dotenv
import logging

# Load .env config
load_dotenv()

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

model_name = os.getenv('AZURE_OPENAI_MODEL')
api_endpoint = os.getenv('AZURE_ENDPOINT')
api_version = os.getenv('AZURE_API_VERSION')
api_key = os.getenv('AZURE_API_KEY')

# Initialize embeddings
embeddings = AzureOpenAIEmbeddings(
    model=model_name,
    openai_api_type="azure",
    azure_endpoint=api_endpoint,
    openai_api_version=api_version,
    openai_api_key=api_key,
)

# Initialize Wikipedia tool
wikipedia = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())

# PostgreSQL connection string
connection_string = "postgresql://talent31stage:centralIndia%402025@talent31-stage-db-centralindia.postgres.database.azure.com:5432/talent31-db-stage-ind?sslmode=require"

# Education level hierarchy for boosting
EDUCATION_BOOST_MAP = {
    'phd': 0.25,
    'doctorate': 0.25,
    'doctoral': 0.25,
    'ph.d': 0.25,
    'ph. d': 0.25,
    'doctor of philosophy': 0.25,
    'master': 0.20,
    'masters': 0.20,
    'msc': 0.20,
    'm.sc': 0.20,
    'm. sc': 0.20,
    'mtech': 0.20,
    'm.tech': 0.20,
    'm. tech': 0.20,
    'mba': 0.20,
    'm.b.a': 0.20,
    'm. b. a': 0.20,
    'me': 0.18,
    'm.e': 0.18,
    'm. e': 0.18,
    'bachelor': 0.10,
    'bachelors': 0.10,
    'bsc': 0.10,
    'b.sc': 0.10,
    'b. sc': 0.10,
    'btech': 0.10,
    'b.tech': 0.10,
    'b. tech': 0.10,
    'be': 0.10,
    'b.e': 0.10,
    'b. e': 0.10,
    'bca': 0.08,
    'b.c.a': 0.08,
    'b. c. a': 0.08,
    'diploma': 0.05,
    'associate': 0.05,
}

def get_wikipedia_context(query, max_length=500):
    """
    Get relevant context from Wikipedia for the given query.
    This can help enhance understanding of technical terms and skills.
    """
    try:
        # Extract key technical terms from the query for Wikipedia search
        technical_terms = []
        common_tech_terms = [
            'javascript', 'typescript', 'react', 'nextjs', 'nodejs', 'python', 
            'java', 'spring', 'kubernetes', 'docker', 'aws', 'azure', 'gcp',
            'machine learning', 'artificial intelligence', 'data science',
            'blockchain', 'devops', 'microservices', 'api', 'database'
        ]
        
        query_lower = query.lower()
        for term in common_tech_terms:
            if term in query_lower:
                technical_terms.append(term)
        
        # If no technical terms found, return empty string
        if not technical_terms:
            logger.info("No technical terms found in query for Wikipedia search.")
            return ""
        
        # Get Wikipedia context for the first relevant term
        primary_term = technical_terms[0]
        try:
            wiki_result = wikipedia.run(primary_term)
            # Truncate to max_length
            if len(wiki_result) > max_length:
                wiki_result = wiki_result[:max_length] + "..."
            
            logger.info(f"Retrieved Wikipedia context for: {primary_term}")
            return f"Wikipedia context for {primary_term}: {wiki_result}"
        except Exception as e:
            logger.warning(f"Could not retrieve Wikipedia context for {primary_term}: {e}")
            return ""
            
    except Exception as e:
        logger.warning(f"Error getting Wikipedia context: {e}")
        return ""

def calculate_education_boost(education_text):
    """
    Calculate education boost score based on the highest degree mentioned.
    Returns the highest boost value found.
    """
    if not education_text:
        return 0.0
    
    education_lower = education_text.lower()
    max_boost = 0.0
    
    for degree, boost in EDUCATION_BOOST_MAP.items():
        if degree in education_lower:
            max_boost = max(max_boost, boost)
    
    return max_boost

def hybrid_search(query, designation=None, skills=None, min_experience=None, preferred_state=None, preferred_country=None, k=10, include_wikipedia=True):
    """
    Performs hybrid search combining semantic similarity with keyword matching boosts,
    education level boosts, and optional Wikipedia context enhancement.
    """
    try:
        # Get Wikipedia context if requested
        wiki_context = ""
        if include_wikipedia:
            wiki_context = get_wikipedia_context(query)
            if wiki_context:
                # Enhance the query with Wikipedia context
                enhanced_query = f"{query}\n\nAdditional Context: {wiki_context}"
            else:
                enhanced_query = query
        else:
            enhanced_query = query
        
        conn = psycopg2.connect(connection_string, cursor_factory=RealDictCursor)
        cursor = conn.cursor()

        # Convert query text to embedding
        query_embedding = embeddings.embed_query(enhanced_query)
        
        # Debug: Check embedding dimension
        logger.info(f"Query embedding dimension: {len(query_embedding)}")
        if wiki_context:
            logger.info(f"Enhanced query with Wikipedia context")
        
        # Convert embedding to PostgreSQL array format
        query_vector_str = "[" + ",".join([str(x) for x in query_embedding]) + "]"

        # Start building the SQL query and parameters
        base_sql = """
        SELECT 
            c."Id",
            c."UserId",
            u."FirstName",
            u."LastName",
            CONCAT(u."FirstName", ' ', u."LastName") as "Name",
            c."DesignationId",
            d."Name" as "DesignationName",
            c."ExperienceYears",
            c."ExperienceMonths",
            c."City",
            c."State",
            c."PostalCode",
            country."Name" as "CountryName",
            c."Summary",
            c."ExpectedCTCFrom",
            c."ExpectedCTCTo",
            c."CurrentCTC",
            c."AdditionalSkills",
            c."LinkedInUrl",
            c."GitHubUrl",
            c."WebsiteUrl",
            c."NoticePeriod",
            c."IsOpenToWork",
            STRING_AGG(DISTINCT proj."Description", ' | ') as "ProjectDescriptions",
            STRING_AGG(DISTINCT exp."Description", ' | ') as "ExperienceDescriptions",
            STRING_AGG(DISTINCT edu."Degree", ' | ') as "EducationDegrees",
            STRING_AGG(DISTINCT CONCAT(edu."Degree", ' from ', edu."Institution"), ' | ') as "EducationDetails",
            1 - (c."Embedding" <=> %s::vector) AS similarity_score,
        """
        
        # Initialize parameters list with the query embedding (used twice)
        params = [query_vector_str, query_vector_str]
        
        # Build boost conditions dynamically
        boost_conditions = ["(1 - (c.\"Embedding\" <=> %s::vector)) * 0.4"]  # Base semantic similarity (40%)
        
        # Add designation boost if provided
        if designation:
            boost_conditions.append("CASE WHEN LOWER(c.\"Summary\") LIKE LOWER(%s) THEN 0.15 ELSE 0 END")
            params.append(f"%{designation.lower()}%")
        
        # Add skills boost if provided
        if skills:
            boost_conditions.append("CASE WHEN LOWER(c.\"AdditionalSkills\") LIKE LOWER(%s) THEN 0.2 ELSE 0 END")
            params.append(f"%{skills.lower()}%")
        
        # Add experience boost if provided
        if min_experience and min_experience > 0:
            boost_conditions.append("CASE WHEN c.\"ExperienceYears\" >= %s THEN 0.08 ELSE 0 END")
            params.append(min_experience)
        
        # Add project description boost
        boost_conditions.append("CASE WHEN EXISTS(SELECT 1 FROM \"CandidateProject\" cp WHERE cp.\"CandidateId\" = c.\"Id\" AND LOWER(cp.\"Description\") LIKE LOWER(%s)) THEN 0.06 ELSE 0 END")
        params.append(f"%{skills.lower() if skills else ''}%")
        
        # Add experience description boost
        boost_conditions.append("CASE WHEN EXISTS(SELECT 1 FROM \"CandidateExperience\" ce WHERE ce.\"CandidateId\" = c.\"Id\" AND LOWER(ce.\"Description\") LIKE LOWER(%s)) THEN 0.05 ELSE 0 END")
        params.append(f"%{skills.lower() if skills else ''}%")
        
        # Add education boost
        education_boost_sql = """
        CASE 
            WHEN EXISTS(
                SELECT 1 FROM "CandidateEducation" ce 
                WHERE ce."CandidateId" = c."Id" 
                AND (
                    LOWER(ce."Degree") LIKE '%phd%' OR LOWER(ce."Degree") LIKE '%doctorate%' OR 
                    LOWER(ce."Degree") LIKE '%doctoral%' OR LOWER(ce."Degree") LIKE '%ph.d%' OR
                    LOWER(ce."Degree") LIKE '%doctor of philosophy%'
                )
            ) THEN 0.25
            WHEN EXISTS(
                SELECT 1 FROM "CandidateEducation" ce 
                WHERE ce."CandidateId" = c."Id" 
                AND (
                    LOWER(ce."Degree") LIKE '%master%' OR LOWER(ce."Degree") LIKE '%msc%' OR 
                    LOWER(ce."Degree") LIKE '%m.sc%' OR LOWER(ce."Degree") LIKE '%mtech%' OR
                    LOWER(ce."Degree") LIKE '%m.tech%' OR LOWER(ce."Degree") LIKE '%mba%' OR
                    LOWER(ce."Degree") LIKE '%m.b.a%' OR LOWER(ce."Degree") LIKE '%me%' OR
                    LOWER(ce."Degree") LIKE '%m.e%'
                )
            ) THEN 0.20
            WHEN EXISTS(
                SELECT 1 FROM "CandidateEducation" ce 
                WHERE ce."CandidateId" = c."Id" 
                AND (
                    LOWER(ce."Degree") LIKE '%bachelor%' OR LOWER(ce."Degree") LIKE '%bsc%' OR 
                    LOWER(ce."Degree") LIKE '%b.sc%' OR LOWER(ce."Degree") LIKE '%btech%' OR
                    LOWER(ce."Degree") LIKE '%b.tech%' OR LOWER(ce."Degree") LIKE '%be%' OR
                    LOWER(ce."Degree") LIKE '%b.e%' OR LOWER(ce."Degree") LIKE '%bca%'
                )
            ) THEN 0.10
            WHEN EXISTS(
                SELECT 1 FROM "CandidateEducation" ce 
                WHERE ce."CandidateId" = c."Id" 
                AND (LOWER(ce."Degree") LIKE '%diploma%' OR LOWER(ce."Degree") LIKE '%associate%')
            ) THEN 0.05
            ELSE 0
        END"""
        boost_conditions.append(education_boost_sql)
        
        # Add state boost (small)
        if preferred_state:
            boost_conditions.append("CASE WHEN LOWER(c.\"State\") LIKE LOWER(%s) THEN 0.08 ELSE 0 END")
            params.append(f"%{preferred_state.lower()}%")
        
        # Add country boost (small)
        if preferred_country:
            boost_conditions.append("CASE WHEN LOWER(country.\"Name\") LIKE LOWER(%s) THEN 0.08 ELSE 0 END")
            params.append(f"%{preferred_country.lower()}%")
        
        # Combine all boost conditions
        boost_sql = " + ".join(boost_conditions)
        
        # Complete the SQL query with joins (including education join)
        complete_sql = base_sql + f"""
            ({boost_sql}) AS boosted_score
        FROM "Candidate" c
        LEFT JOIN "User" u ON c."UserId" = u."Id"
        LEFT JOIN "Designation" d ON c."DesignationId" = d."Id"
        LEFT JOIN "Country" country ON c."CountryId" = country."Id"
        LEFT JOIN "CandidateProject" proj ON c."Id" = proj."CandidateId"
        LEFT JOIN "CandidateExperience" exp ON c."Id" = exp."CandidateId"
        LEFT JOIN "CandidateEducation" edu ON c."Id" = edu."CandidateId"
        WHERE c."Embedding" IS NOT NULL
        GROUP BY c."Id", u."FirstName", u."LastName", d."Name", country."Name",
                 c."UserId", c."DesignationId", c."ExperienceYears", c."ExperienceMonths",
                 c."City", c."State", c."PostalCode", c."Summary", c."ExpectedCTCFrom",
                 c."ExpectedCTCTo", c."CurrentCTC", c."AdditionalSkills", c."LinkedInUrl",
                 c."GitHubUrl", c."WebsiteUrl", c."NoticePeriod", c."IsOpenToWork", c."Embedding"
        ORDER BY boosted_score DESC
        LIMIT %s;
        """
        
        # Add the limit parameter
        params.append(k)
        
        # Debug: Print SQL and parameter count
        logger.info(f"SQL placeholders count: {complete_sql.count('%s')}")
        logger.info(f"Parameters count: {len(params)}")
        
        # Execute the query
        cursor.execute(complete_sql, params)
        results = cursor.fetchall()
        
        logger.info(f"Found {len(results)} candidates")
        
        # Add Wikipedia context to results metadata if used
        if wiki_context:
            for result in results:
                result['wikipedia_context_used'] = True
                result['wikipedia_context'] = wiki_context[:200] + "..." if len(wiki_context) > 200 else wiki_context
        
        return results
        
    except Exception as e:
        logger.error(f"Error in hybrid_search: {str(e)}")
        raise
    finally:
        if 'conn' in locals():
            conn.close()


if __name__ == "__main__":
    print("=== Testing Enhanced Hybrid Search with Education Boost ===")
    try:
        query = """We are looking for a Frontend Developer (India-based) with 3–5 years of experience.
        Must-have skills: JavaScript, TypeScript, React.js, Next.js, HTML5, CSS3, TailwindCSS, Redux, RESTful APIs, GraphQL, Webpack, Babel, Jest/React Testing Library, Cypress, Git/GitHub, CI/CD pipelines, Docker, Kubernetes, AWS (CloudFront, S3), Figma/Adobe XD, Responsive Web Design, Agile/Scrum methodologies"""
                
        designation = "Frontend Developer"
        skills = "JavaScript, TypeScript, React.js, Next.js, HTML5, CSS3, TailwindCSS, Redux, RESTful APIs, GraphQL, Webpack, Babel, Jest, Cypress, Git, CI/CD, Docker, Kubernetes, AWS CloudFront, AWS S3, Figma, Adobe XD, Responsive Web Design, Agile"
        min_experience = 5
        preferred_state = "Karnataka"
        preferred_country = "India"

        candidates = hybrid_search(
            query, 
            designation, 
            skills, 
            min_experience, 
            preferred_state, 
            preferred_country, 
            k=10,
            include_wikipedia=True
        )

        print(f"Enhanced hybrid search found {len(candidates)} results")
        
        for i, c in enumerate(candidates, 1):
            print(f"\n--- Enhanced Candidate {i} ---")
            print(f"ID: {c['Id']}")
            print(f"Name: {c.get('Name', 'N/A')}")
            print(f"Designation: {c.get('DesignationName', 'N/A')}")
            print(f"Boosted Score: {c['boosted_score']:.3f}")
            print(f"Similarity Score: {c['similarity_score']:.3f}")
            print(f"Experience: {c['ExperienceYears']} years {c.get('ExperienceMonths', 0)} months")
            print(f"Location: {c.get('City', 'N/A')}, {c.get('State', 'N/A')}, {c.get('CountryName', 'N/A')}")
            print(f"Expected CTC: {c.get('ExpectedCTCFrom', 'N/A')} - {c.get('ExpectedCTCTo', 'N/A')}")
            print(f"Current CTC: {c.get('CurrentCTC', 'N/A')}")
            print(f"Notice Period: {c.get('NoticePeriod', 'N/A')}")
            print(f"Open to Work: {c.get('IsOpenToWork', 'N/A')}")
            print(f"Skills: {c.get('AdditionalSkills', 'No skills listed')}")
            
            # Display education information
            if c.get('EducationDegrees'):
                print(f"Education: {c['EducationDegrees']}")
            if c.get('EducationDetails'):
                print(f"Education Details: {c['EducationDetails']}")
            
            print(f"Summary: {c['Summary'][:150] if c.get('Summary') else 'No summary'}...")
            
            # Display project descriptions if available
            if c.get('ProjectDescriptions'):
                print(f"Project Experience: {c['ProjectDescriptions'][:200]}...")
            
            # Display work experience descriptions if available  
            if c.get('ExperienceDescriptions'):
                print(f"Work Experience: {c['ExperienceDescriptions'][:200]}...")
                
            # Display Wikipedia context if used
            if c.get('wikipedia_context_used'):
                print(f"Wikipedia Enhanced: Yes")
                print(f"Context: {c.get('wikipedia_context', '')}")
                
            if c.get('LinkedInUrl'):
                print(f"LinkedIn: {c['LinkedInUrl']}")
            if c.get('GitHubUrl'):
                print(f"GitHub: {c['GitHubUrl']}")
            if c.get('WebsiteUrl'):
                print(f"Website: {c['WebsiteUrl']}")
            
        # Test education boost calculation separately
        print("\n=== Education Boost Examples ===")
        test_educations = [
            "Master of Science in Computer Science",
            "Bachelor of Technology in Information Technology", 
            "PhD in Artificial Intelligence",
            "MBA from IIM",
            "Diploma in Computer Applications",
            "B.Tech from IIT Delhi, M.Tech from IIT Bombay"
        ]
        
        for edu in test_educations:
            boost = calculate_education_boost(edu)
            print(f"Education: {edu} -> Boost: {boost}")
            
    except Exception as e:
        print(f"Enhanced hybrid search failed: {e}")
        import traceback
        traceback.print_exc()
```

```python
import os
import psycopg2
from psycopg2.extras import RealDictCursor
from langchain_openai import AzureOpenAIEmbeddings
from dotenv import load_dotenv
import logging
from datetime import datetime, timedelta

# Load .env config
load_dotenv()

# Set up logging
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger(__name__)

model_name = os.getenv('AZURE_OPENAI_MODEL')
api_endpoint = os.getenv('AZURE_ENDPOINT')
api_version = os.getenv('AZURE_API_VERSION')
api_key = os.getenv('AZURE_API_KEY')

# Initialize embeddings
embeddings = AzureOpenAIEmbeddings(
    model=model_name,
    openai_api_type="azure",
    azure_endpoint=api_endpoint,
    openai_api_version=api_version,
    openai_api_key=api_key,
)

# PostgreSQL connection string
connection_string = "postgresql://talent31stage:centralIndia%402025@talent31-stage-db-centralindia.postgres.database.azure.com:5432/talent31-db-stage-ind?sslmode=require"

def improved_hybrid_search(query, designation=None, skills=None, min_experience=None, 
                         max_experience=None, preferred_state=None, preferred_country=None, 
                         min_ctc=None, max_ctc=None, max_notice_period=None, 
                         education_level=None, company_type=None, k=10):
    """
    Enhanced hybrid search with predictive analytics and comprehensive candidate profiling.
    Incorporates modern AI-driven talent matching principles for better candidate-job alignment.
    """
    try:
        conn = psycopg2.connect(connection_string, cursor_factory=RealDictCursor)
        cursor = conn.cursor()

        # Convert query text to embedding
        query_embedding = embeddings.embed_query(query)
        
        # Debug: Check embedding dimension
        logger.info(f"Query embedding dimension: {len(query_embedding)}")
        
        # Convert embedding to PostgreSQL array format
        query_vector_str = "[" + ",".join([str(x) for x in query_embedding]) + "]"

        # Enhanced SQL with predictive columns and analytics
        base_sql = """
        SELECT 
            c."Id",
            c."UserId",
            u."FirstName",
            u."LastName",
            CONCAT(u."FirstName", ' ', u."LastName") as "Name",
            c."DesignationId",
            d."Name" as "DesignationName",
            c."ExperienceYears",
            c."ExperienceMonths",
            c."City",
            c."State",
            c."PostalCode",
            country."Name" as "CountryName",
            c."Summary",
            c."ExpectedCTCFrom",
            c."ExpectedCTCTo",
            c."CurrentCTC",
            c."PayType",
            curr."Name" as "CurrencyName",
            c."AdditionalSkills",
            c."LinkedInUrl",
            c."GitHubUrl",
            c."WebsiteUrl",
            c."NoticePeriod",
            c."IsOpenToWork",
            c."OnBench",
            c."SubscriptionType",
            u."DOB",
            u."LastLoginAt",
            u."CreatedAt" as "UserCreatedAt",
            
            -- Aggregated education data
            STRING_AGG(DISTINCT CONCAT(ce."Degree", ' from ', ce."Institution", ' (', ce."GraduationYear", ')'), ' | ') as "EducationDetails",
            MAX(ce."GraduationYear") as "LatestGraduation",
            COUNT(DISTINCT ce."Id") as "EducationCount",
            
            -- Enhanced experience aggregation with analytics
            STRING_AGG(DISTINCT exp."CompanyName", ', ') as "PreviousCompanies",
            STRING_AGG(DISTINCT exp."JobTitle", ' | ') as "PreviousRoles",
            STRING_AGG(DISTINCT exp."Description", ' | ') as "ExperienceDescriptions",
            COUNT(DISTINCT exp."Id") as "NumberOfJobs",
            
            -- Calculate average job tenure (handling null end dates as current jobs)
            AVG(
                CASE 
                    WHEN exp."EndDate" IS NOT NULL THEN 
                        EXTRACT(EPOCH FROM (exp."EndDate" - exp."StartDate")) / (365.25 * 24 * 3600)
                    ELSE 
                        EXTRACT(EPOCH FROM (NOW() - exp."StartDate")) / (365.25 * 24 * 3600)
                END
            ) as "AvgJobTenure",
            
            -- Most recent job end date
            MAX(exp."EndDate") as "LastJobEndDate",
            MIN(exp."StartDate") as "FirstJobStartDate",
            
            -- Project experience analytics
            STRING_AGG(DISTINCT proj."Name", ' | ') as "ProjectNames",
            STRING_AGG(DISTINCT proj."Role", ' | ') as "ProjectRoles",
            STRING_AGG(DISTINCT proj."Description", ' | ') as "ProjectDescriptions",
            COUNT(DISTINCT proj."Id") as "NumberOfProjects",
            
            -- Calculate project duration experience
            AVG(
                CASE 
                    WHEN proj."EndDate" IS NOT NULL THEN 
                        EXTRACT(EPOCH FROM (proj."EndDate" - proj."StartDate")) / (30.44 * 24 * 3600)
                    ELSE NULL
                END
            ) as "AvgProjectDuration",
            
            -- Skill proficiency data with enhanced analytics
            STRING_AGG(DISTINCT CONCAT(s."Name", ' (', COALESCE(cs."MinimumExperience", 0), ' years)'), ' | ') as "DetailedSkills",
            COUNT(DISTINCT cs."SkillId") as "SkillCount",
            AVG(COALESCE(cs."MinimumExperience", 0)) as "AvgSkillExperience",
            COUNT(DISTINCT CASE WHEN s."IsTechnical" = true THEN cs."SkillId" END) as "TechnicalSkillCount",
            
            -- Engagement and activity metrics
            CASE 
                WHEN u."LastLoginAt" > NOW() - INTERVAL '7 days' THEN 1.0
                WHEN u."LastLoginAt" > NOW() - INTERVAL '30 days' THEN 0.8
                WHEN u."LastLoginAt" > NOW() - INTERVAL '90 days' THEN 0.5
                WHEN u."LastLoginAt" > NOW() - INTERVAL '180 days' THEN 0.3
                ELSE 0.1
            END as "EngagementScore",
            
            -- Profile completeness score
            (
                CASE WHEN c."Summary" IS NOT NULL AND LENGTH(c."Summary") > 50 THEN 0.2 ELSE 0 END +
                CASE WHEN c."AdditionalSkills" IS NOT NULL AND LENGTH(c."AdditionalSkills") > 10 THEN 0.15 ELSE 0 END +
                CASE WHEN c."LinkedInUrl" IS NOT NULL THEN 0.15 ELSE 0 END +
                CASE WHEN c."ExpectedCTCFrom" IS NOT NULL THEN 0.1 ELSE 0 END +
                CASE WHEN c."NoticePeriod" IS NOT NULL THEN 0.1 ELSE 0 END +
                CASE WHEN COUNT(DISTINCT exp."Id") > 0 THEN 0.15 ELSE 0 END +
                CASE WHEN COUNT(DISTINCT ce."Id") > 0 THEN 0.1 ELSE 0 END +
                CASE WHEN COUNT(DISTINCT proj."Id") > 0 THEN 0.05 ELSE 0 END
            ) as "ProfileCompletenessScore",
            
            -- Age calculation
            EXTRACT(YEAR FROM AGE(COALESCE(u."DOB", NOW() - INTERVAL '30 years'))) as "Age",
            
            -- Career progression indicators
            CASE 
                WHEN c."ExperienceYears" > 0 AND COUNT(DISTINCT exp."Id") > 0 THEN 
                    COUNT(DISTINCT exp."Id")::float / NULLIF(c."ExperienceYears", 0)
                ELSE 0
            END as "CareerProgressionRate",
            
            -- Employment gap analysis
            CASE 
                WHEN MAX(exp."EndDate") IS NOT NULL THEN
                    EXTRACT(EPOCH FROM (NOW() - MAX(exp."EndDate"))) / (30.44 * 24 * 3600)
                ELSE 0
            END as "MonthsSinceLastJob",
            
            -- Salary growth potential
            CASE 
                WHEN c."CurrentCTC" IS NOT NULL AND c."ExpectedCTCFrom" IS NOT NULL AND c."CurrentCTC" > 0 THEN
                    (c."ExpectedCTCFrom" - c."CurrentCTC")::float / c."CurrentCTC" * 100
                ELSE NULL
            END as "ExpectedSalaryGrowthPercent",
            
            -- Base similarity score
            1 - (c."Embedding" <=> %s::vector) AS similarity_score,
        """
        
        # Initialize parameters list with the query embedding (used twice)
        params = [query_vector_str, query_vector_str]
        
        # Enhanced boost conditions with predictive factors
        boost_conditions = ["(1 - (c.\"Embedding\" <=> %s::vector)) * 0.35"]  # Base semantic similarity (35%)
        
        # Designation matching boost (15% max)
        if designation:
            boost_conditions.append("""
                CASE 
                    WHEN LOWER(d."Name") = LOWER(%s) THEN 0.15
                    WHEN LOWER(d."Name") LIKE LOWER(%s) THEN 0.12
                    WHEN LOWER(c."Summary") LIKE LOWER(%s) THEN 0.08
                    ELSE 0 
                END
            """)
            params.extend([designation.lower(), f"%{designation.lower()}%", f"%{designation.lower()}%"])
        
        # Enhanced skills matching boost (20% max)
        if skills:
            skill_list = [skill.strip().lower() for skill in skills.split(',')]
            skill_conditions = []
            
            for skill in skill_list[:5]:  # Limit to top 5 skills for performance
                skill_conditions.append(f"LOWER(c.\"AdditionalSkills\") LIKE LOWER(%s)")
                params.append(f"%{skill}%")
            
            if skill_conditions:
                boost_conditions.append(f"""
                    CASE WHEN ({' OR '.join(skill_conditions)}) THEN 0.2
                         WHEN EXISTS(SELECT 1 FROM "CandidateSkill" cs2 
                                   JOIN "Skill" s2 ON cs2."SkillId" = s2."Id" 
                                   WHERE cs2."CandidateId" = c."Id" 
                                   AND ({' OR '.join([f'LOWER(s2."Name") LIKE LOWER(%s)' for _ in skill_list[:5]])}))
                         THEN 0.15
                         ELSE 0 
                    END
                """)
                params.extend([f"%{skill}%" for skill in skill_list[:5]])
        
        # Experience range boost with nuanced scoring (15% max)
        if min_experience and max_experience:
            boost_conditions.append("""
                CASE 
                    WHEN c."ExperienceYears" BETWEEN %s AND %s THEN 0.15
                    WHEN c."ExperienceYears" BETWEEN %s - 1 AND %s + 1 THEN 0.12
                    WHEN c."ExperienceYears" BETWEEN %s - 2 AND %s + 2 THEN 0.08
                    ELSE 0 
                END
            """)
            params.extend([min_experience, max_experience, min_experience, max_experience, min_experience, max_experience])
        elif min_experience:
            boost_conditions.append("""
                CASE 
                    WHEN c."ExperienceYears" >= %s THEN 0.12
                    WHEN c."ExperienceYears" >= %s - 1 THEN 0.08
                    ELSE 0 
                END
            """)
            params.extend([min_experience, min_experience])
        
        # Salary expectation alignment boost (12% max)
        if min_ctc and max_ctc:
            boost_conditions.append("""
                CASE 
                    WHEN c."ExpectedCTCFrom" <= %s AND c."ExpectedCTCTo" >= %s THEN 0.12
                    WHEN c."ExpectedCTCFrom" <= %s * 1.15 AND c."ExpectedCTCTo" >= %s * 0.85 THEN 0.09
                    WHEN c."ExpectedCTCFrom" <= %s * 1.3 AND c."ExpectedCTCTo" >= %s * 0.7 THEN 0.06
                    ELSE 0 
                END
            """)
            params.extend([max_ctc, min_ctc, max_ctc, min_ctc, max_ctc, min_ctc])
        
        # Notice period compatibility boost (10% max)
        if max_notice_period:
            boost_conditions.append("""
                CASE 
                    WHEN c."NoticePeriod" <= %s THEN 0.1
                    WHEN c."NoticePeriod" <= %s * 1.5 THEN 0.06
                    ELSE 0 
                END
            """)
            params.extend([max_notice_period, max_notice_period])
        
        # Availability and engagement boosts
        boost_conditions.extend([
            # Open to work status (8% max)
            "CASE WHEN c.\"IsOpenToWork\" = true THEN 0.08 ELSE 0 END",
            
            # On bench status (5% max)
            "CASE WHEN c.\"OnBench\" = true THEN 0.05 ELSE 0 END",
            
            # Recent activity boost (8% max)
            """CASE 
                WHEN u."LastLoginAt" > NOW() - INTERVAL '7 days' THEN 0.08
                WHEN u."LastLoginAt" > NOW() - INTERVAL '30 days' THEN 0.06
                WHEN u."LastLoginAt" > NOW() - INTERVAL '90 days' THEN 0.04
                WHEN u."LastLoginAt" > NOW() - INTERVAL '180 days' THEN 0.02
                ELSE 0 
            END""",
            
            # Profile completeness boost (6% max)
            """(
                CASE WHEN c."Summary" IS NOT NULL AND LENGTH(c."Summary") > 50 THEN 0.015 ELSE 0 END +
                CASE WHEN c."AdditionalSkills" IS NOT NULL AND LENGTH(c."AdditionalSkills") > 10 THEN 0.01 ELSE 0 END +
                CASE WHEN c."LinkedInUrl" IS NOT NULL THEN 0.01 ELSE 0 END +
                CASE WHEN c."ExpectedCTCFrom" IS NOT NULL THEN 0.005 ELSE 0 END +
                CASE WHEN COUNT(DISTINCT exp."Id") > 0 THEN 0.015 ELSE 0 END +
                CASE WHEN COUNT(DISTINCT ce."Id") > 0 THEN 0.005 ELSE 0 END
            )"""
        ])
        
        # Education level boost (6% max)
        if education_level:
            boost_conditions.append("""
                CASE WHEN EXISTS(
                    SELECT 1 FROM "CandidateEducation" ce2 
                    WHERE ce2."CandidateId" = c."Id" 
                    AND (LOWER(ce2."Degree") LIKE LOWER(%s) OR LOWER(ce2."Degree") LIKE LOWER(%s))
                ) THEN 0.06 ELSE 0 END
            """)
            params.extend([f"%{education_level.lower()}%", f"%{education_level.lower().replace('bachelor', 'b.')}%"])
        
        # Geographic preference boosts
        if preferred_state:
            boost_conditions.append("""
                CASE 
                    WHEN LOWER(c."State") = LOWER(%s) THEN 0.07
                    WHEN LOWER(c."State") LIKE LOWER(%s) THEN 0.05
                    ELSE 0 
                END
            """)
            params.extend([preferred_state.lower(), f"%{preferred_state.lower()}%"])
        
        if preferred_country:
            boost_conditions.append("""
                CASE 
                    WHEN LOWER(country."Name") = LOWER(%s) THEN 0.05
                    WHEN LOWER(country."Name") LIKE LOWER(%s) THEN 0.03
                    ELSE 0 
                END
            """)
            params.extend([preferred_country.lower(), f"%{preferred_country.lower()}%"])
        
        # Career stability and progression boosts (4% max each)
        boost_conditions.extend([
            # Job stability boost
            """CASE 
                WHEN AVG(
                    CASE 
                        WHEN exp."EndDate" IS NOT NULL THEN 
                            EXTRACT(EPOCH FROM (exp."EndDate" - exp."StartDate")) / (365.25 * 24 * 3600)
                        ELSE 
                            EXTRACT(EPOCH FROM (NOW() - exp."StartDate")) / (365.25 * 24 * 3600)
                    END
                ) >= 2.5 THEN 0.04
                WHEN AVG(
                    CASE 
                        WHEN exp."EndDate" IS NOT NULL THEN 
                            EXTRACT(EPOCH FROM (exp."EndDate" - exp."StartDate")) / (365.25 * 24 * 3600)
                        ELSE 
                            EXTRACT(EPOCH FROM (NOW() - exp."StartDate")) / (365.25 * 24 * 3600)
                    END
                ) >= 1.5 THEN 0.02
                ELSE 0 
            END""",
            
            # Project experience boost
            """CASE 
                WHEN COUNT(DISTINCT proj."Id") >= 5 THEN 0.04
                WHEN COUNT(DISTINCT proj."Id") >= 3 THEN 0.03
                WHEN COUNT(DISTINCT proj."Id") >= 1 THEN 0.02
                ELSE 0 
            END""",
            
            # Age and experience balance boost
            """CASE 
                WHEN EXTRACT(YEAR FROM AGE(COALESCE(u."DOB", NOW() - INTERVAL '30 years'))) BETWEEN 25 AND 35 
                     AND c."ExperienceYears" BETWEEN 3 AND 8 THEN 0.03
                WHEN EXTRACT(YEAR FROM AGE(COALESCE(u."DOB", NOW() - INTERVAL '30 years'))) BETWEEN 28 AND 40 
                     AND c."ExperienceYears" BETWEEN 5 AND 12 THEN 0.02
                ELSE 0 
            END""",
            
            # Employment gap penalty (negative boost)
            """CASE 
                WHEN MAX(exp."EndDate") IS NOT NULL AND 
                     EXTRACT(EPOCH FROM (NOW() - MAX(exp."EndDate"))) / (30.44 * 24 * 3600) > 12 THEN -0.02
                WHEN MAX(exp."EndDate") IS NOT NULL AND 
                     EXTRACT(EPOCH FROM (NOW() - MAX(exp."EndDate"))) / (30.44 * 24 * 3600) > 6 THEN -0.01
                ELSE 0 
            END""",
            
            # Skill diversity boost
            """CASE 
                WHEN COUNT(DISTINCT cs."SkillId") >= 8 THEN 0.02
                WHEN COUNT(DISTINCT cs."SkillId") >= 5 THEN 0.01
                ELSE 0 
            END"""
        ])
        
        # Combine all boost conditions
        boost_sql = " + ".join(boost_conditions)
        
        # Complete SQL with enhanced joins and grouping
        complete_sql = base_sql + f"""
            ({boost_sql}) AS boosted_score
        FROM "Candidate" c
        LEFT JOIN "User" u ON c."UserId" = u."Id"
        LEFT JOIN "Designation" d ON c."DesignationId" = d."Id"
        LEFT JOIN "Country" country ON c."CountryId" = country."Id"
        LEFT JOIN "Currency" curr ON c."CurrencyId" = curr."Id"
        LEFT JOIN "CandidateEducation" ce ON c."Id" = ce."CandidateId"
        LEFT JOIN "CandidateExperience" exp ON c."Id" = exp."CandidateId"
        LEFT JOIN "CandidateProject" proj ON c."Id" = proj."CandidateId"
        LEFT JOIN "CandidateSkill" cs ON c."Id" = cs."CandidateId"
        LEFT JOIN "Skill" s ON cs."SkillId" = s."Id"
        WHERE c."Embedding" IS NOT NULL
        GROUP BY c."Id", u."FirstName", u."LastName", d."Name", country."Name", curr."Name",
                 c."UserId", c."DesignationId", c."ExperienceYears", c."ExperienceMonths",
                 c."City", c."State", c."PostalCode", c."Summary", c."ExpectedCTCFrom",
                 c."ExpectedCTCTo", c."CurrentCTC", c."PayType", c."AdditionalSkills", 
                 c."LinkedInUrl", c."GitHubUrl", c."WebsiteUrl", c."NoticePeriod", 
                 c."IsOpenToWork", c."OnBench", c."SubscriptionType", u."DOB", 
                 u."LastLoginAt", u."CreatedAt", c."Embedding"
        ORDER BY boosted_score DESC
        LIMIT %s;
        """
        
        # Add the limit parameter
        params.append(k)
        
        # Debug information
        logger.info(f"SQL placeholders count: {complete_sql.count('%s')}")
        logger.info(f"Parameters count: {len(params)}")
        
        # Execute the query
        cursor.execute(complete_sql, params)
        results = cursor.fetchall()
        
        logger.info(f"Enhanced hybrid search found {len(results)} candidates")
        
        return results
        
    except Exception as e:
        logger.error(f"Error in improved_hybrid_search: {str(e)}")
        raise
    finally:
        if 'conn' in locals():
            conn.close()

def analyze_candidate_metrics(candidate_data):
    """
    Analyze and provide insights on candidate metrics for better decision making.
    """
    insights = {
        'engagement_level': 'Unknown',
        'career_stability': 'Unknown',
        'profile_quality': 'Unknown',
        'skill_diversity': 'Unknown',
        'experience_relevance': 'Unknown'
    }
    
    # Engagement analysis
    engagement_score = candidate_data.get('EngagementScore', 0)
    if engagement_score >= 0.8:
        insights['engagement_level'] = 'Highly Active'
    elif engagement_score >= 0.5:
        insights['engagement_level'] = 'Moderately Active'
    elif engagement_score >= 0.3:
        insights['engagement_level'] = 'Occasionally Active'
    else:
        insights['engagement_level'] = 'Inactive'
    
    # Career stability analysis
    avg_tenure = candidate_data.get('AvgJobTenure', 0) or 0
    if avg_tenure >= 2.5:
        insights['career_stability'] = 'Very Stable'
    elif avg_tenure >= 1.5:
        insights['career_stability'] = 'Stable'
    elif avg_tenure >= 1:
        insights['career_stability'] = 'Moderate'
    else:
        insights['career_stability'] = 'High Turnover'
    
    # Profile completeness analysis
    completeness_score = candidate_data.get('ProfileCompletenessScore', 0) or 0
    if completeness_score >= 0.8:
        insights['profile_quality'] = 'Excellent'
    elif completeness_score >= 0.6:
        insights['profile_quality'] = 'Good'
    elif completeness_score >= 0.4:
        insights['profile_quality'] = 'Fair'
    else:
        insights['profile_quality'] = 'Poor'
    
    # Skill diversity analysis
    skill_count = candidate_data.get('SkillCount', 0) or 0
    technical_skills = candidate_data.get('TechnicalSkillCount', 0) or 0
    
    if skill_count >= 8 and technical_skills >= 5:
        insights['skill_diversity'] = 'Highly Diverse'
    elif skill_count >= 5 and technical_skills >= 3:
        insights['skill_diversity'] = 'Well Rounded'
    elif skill_count >= 3:
        insights['skill_diversity'] = 'Focused'
    else:
        insights['skill_diversity'] = 'Limited'
    
    # Experience relevance
    boosted_score = candidate_data.get('boosted_score', 0) or 0
    if boosted_score >= 0.8:
        insights['experience_relevance'] = 'Excellent Match'
    elif boosted_score >= 0.65:
        insights['experience_relevance'] = 'Good Match'
    elif boosted_score >= 0.5:
        insights['experience_relevance'] = 'Fair Match'
    else:
        insights['experience_relevance'] = 'Poor Match'
    
    return insights

def format_candidate_output(candidate, index, include_insights=True):
    """
    Format candidate output with enhanced analytics and insights.
    """
    output = [f"\n--- Enhanced Candidate {index} ---"]
    output.append(f"ID: {candidate['Id']}")
    output.append(f"Name: {candidate.get('Name', 'N/A')}")
    output.append(f"Designation: {candidate.get('DesignationName', 'N/A')}")
    
    # Scoring metrics
    output.append(f"Overall Match Score: {candidate['boosted_score']:.3f}")
    output.append(f"Semantic Similarity: {candidate['similarity_score']:.3f}")
    output.append(f"Engagement Score: {candidate.get('EngagementScore', 0):.3f}")
    output.append(f"Profile Completeness: {candidate.get('ProfileCompletenessScore', 0):.3f}")
    
    # Basic info
    experience_months = candidate.get('ExperienceMonths', 0) or 0
    output.append(f"Experience: {candidate['ExperienceYears']} years {experience_months} months")
    output.append(f"Age: {candidate.get('Age', 'N/A')} years")
    output.append(f"Location: {candidate.get('City', 'N/A')}, {candidate.get('State', 'N/A')}, {candidate.get('CountryName', 'N/A')}")
    
    # Financial details
    currency = candidate.get('CurrencyName', '')
    output.append(f"Expected CTC: {candidate.get('ExpectedCTCFrom', 'N/A')} - {candidate.get('ExpectedCTCTo', 'N/A')} {currency}")
    output.append(f"Current CTC: {candidate.get('CurrentCTC', 'N/A')} {currency}")
    
    if candidate.get('ExpectedSalaryGrowthPercent'):
        output.append(f"Expected Growth: {candidate['ExpectedSalaryGrowthPercent']:.1f}%")
    
    # Availability
    output.append(f"Notice Period: {candidate.get('NoticePeriod', 'N/A')} days")
    output.append(f"Open to Work: {candidate.get('IsOpenToWork', 'N/A')}")
    output.append(f"On Bench: {candidate.get('OnBench', 'N/A')}")
    
    # Career analytics
    if candidate.get('AvgJobTenure'):
        output.append(f"Average Job Tenure: {candidate['AvgJobTenure']:.1f} years")
    
    output.append(f"Number of Jobs: {candidate.get('NumberOfJobs', 0)}")
    output.append(f"Number of Projects: {candidate.get('NumberOfProjects', 0)}")
    output.append(f"Skill Count: {candidate.get('SkillCount', 0)} ({candidate.get('TechnicalSkillCount', 0)} technical)")
    
    if candidate.get('CareerProgressionRate'):
        output.append(f"Career Progression Rate: {candidate['CareerProgressionRate']:.2f} jobs/year")
    
    if candidate.get('MonthsSinceLastJob') and candidate['MonthsSinceLastJob'] > 0:
        output.append(f"Months Since Last Job: {candidate['MonthsSinceLastJob']:.1f}")
    
    # Skills and experience
    output.append(f"Skills: {candidate.get('AdditionalSkills', 'No skills listed')}")
    
    if candidate.get('DetailedSkills'):
        output.append(f"Detailed Skills: {candidate['DetailedSkills'][:200]}...")
    
    output.append(f"Summary: {candidate['Summary'][:200] if candidate.get('Summary') else 'No summary'}...")
    
    # Education
    if candidate.get('EducationDetails'):
        output.append(f"Education: {candidate['EducationDetails']}")
    
    # Professional experience
    if candidate.get('PreviousCompanies'):
        output.append(f"Previous Companies: {candidate['PreviousCompanies']}")
    
    if candidate.get('PreviousRoles'):
        output.append(f"Previous Roles: {candidate['PreviousRoles'][:150]}...")
    
    if candidate.get('ExperienceDescriptions'):
        output.append(f"Work Experience: {candidate['ExperienceDescriptions'][:200]}...")
    
    if candidate.get('ProjectDescriptions'):
        output.append(f"Project Experience: {candidate['ProjectDescriptions'][:200]}...")
    
    # Contact info
    if candidate.get('LinkedInUrl'):
        output.append(f"LinkedIn: {candidate['LinkedInUrl']}")
    if candidate.get('GitHubUrl'):
        output.append(f"GitHub: {candidate['GitHubUrl']}")
    if candidate.get('WebsiteUrl'):
        output.append(f"Website: {candidate['WebsiteUrl']}")
    
    # AI-driven insights
    if include_insights:
        insights = analyze_candidate_metrics(candidate)
        output.append(f"\n📊 AI Insights:")
        output.append(f"  • Engagement Level: {insights['engagement_level']}")
        output.append(f"  • Career Stability: {insights['career_stability']}")
        output.append(f"  • Profile Quality: {insights['profile_quality']}")
        output.append(f"  • Skill Diversity: {insights['skill_diversity']}")
        output.append(f"  • Experience Relevance: {insights['experience_relevance']}")
    
    return "\n".join(output)

if __name__ == "__main__":
    print("=== Enhanced AI-Powered Hybrid Search ===")
    try:
        # Enhanced search query
        query = """We are looking for a Senior Backend + AWS Developer (India-based) with 3–7 years of experience.
Must-have skills: Python/Java/Node.js, AWS (EC2, S3, Lambda, RDS, CloudFormation), RESTful APIs, 
Microservices Architecture, Docker, Kubernetes, PostgreSQL/MySQL, Redis, Git/GitHub, CI/CD pipelines, 
Terraform, API Gateway, CloudWatch, ElasticSearch, Message queues (SQS/SNS), Linux/Unix.
Nice-to-have: GraphQL, MongoDB, Kafka, Prometheus, Grafana, Helm charts.
"""
                
        # Enhanced search parameters
        search_params = {
            "query": query,
            "designation": "Backend Developer",
            "skills": "Python, Java, Node.js, AWS, Docker, Kubernetes, PostgreSQL, Redis, Git, CI/CD, Terraform, API Gateway, CloudWatch, ElasticSearch, SQS, SNS, Linux, Unix",
            "min_experience": 3,
            "max_experience": 7,
            "min_ctc": 800000,  # 8 LPA
            "max_ctc": 2500000,  # 25 LPA
            "max_notice_period": 90,  # 3 months
            "education_level": "Bachelor",
            "preferred_country": "India",
            "k": 15
        }

        print(f"🔍 Searching with parameters:")
        for key, value in search_params.items():
            if key != "query":
                print(f"  • {key}: {value}")
        
        print(f"\n📝 Job Description:\n{query}")

        # Execute enhanced search
        candidates = improved_hybrid_search(**search_params)

        print(f"\n🎯 Enhanced hybrid search found {len(candidates)} results")
        print("=" * 80)
        
        # Display results with insights
        for i, candidate in enumerate(candidates, 1):
            print(format_candidate_output(candidate, i, include_insights=True))
            
            if i < len(candidates):
                print("\n" + "="*80)
        
        # Summary analytics
        if candidates:
            print(f"\n📈 Search Summary:")
            avg_score = sum(c['boosted_score'] for c in candidates) / len(candidates)
            avg_engagement = sum(c.get('EngagementScore', 0) for c in candidates) / len(candidates)
            active_candidates = sum(1 for c in candidates if c.get('IsOpenToWork'))
            
            print(f"  • Average Match Score: {avg_score:.3f}")
            print(f"  • Average Engagement: {avg_engagement:.3f}")
            print(f"  • Candidates Open to Work: {active_candidates}/{len(candidates)}")
            print(f"  • Top Match Score: {candidates[0]['boosted_score']:.3f}")
            print(f"  • Score Range: {candidates[-1]['boosted_score']:.3f} - {candidates[0]['boosted_score']:.3f}")
            
    except Exception as e:
        print(f"Enhanced hybrid search failed: {e}")
        import traceback
        traceback.print_exc()
```
